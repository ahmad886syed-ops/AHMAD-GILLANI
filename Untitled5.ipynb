```python
import pandas as pd
import numpy as np

df=pd.DataFrame({
 'Name':['Alice','Bob','Charlie','Alice',] ,
 'Age':[25,np.nan,35,25],
 'Salary':['50000','60000',np.nan,'50000']})

df=df.drop_duplicates()
df['Age'].fillna(df['Age'].mean(),inplace=True)
df['Salary'].fillna(df['Salary'].mean(),inplace=True)
```
